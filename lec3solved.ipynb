#  MULTIDIMENSIONAL ARRAYS

```python
#Create an array with given "layers"
#The first one should be the identity matrix,
#the second one the matrix of ones,
#the third one should be -3 times the second one 
#check its shape
#change element at (2,3) in the last layer to be 5
A1 = np.eye(4)
print("Identity matrix:")
print(A1)
A2 = np.ones((4,4))
print(" Matrix of ones:")
print(A2)
A3 = -3*A2
print(" -3 * Matrix of ones:")
print(A3)
print(" Assembling a multidimensional array containing these 3 matrices")
A_mult = np.array([A1, A2, A3])
print(A_mult)
print("Its shape is :")
print(A_mult.shape)

# Get a specific element
print("How to access the elements of this array?")
print(f'Obtaining the element (3,2) of the third matrix : {A_mult[2,1,2]} ')


#Change it to be 5
A_mult[2,1,2] = 5

# Print the new one
print("we have changed this number from -3 to 5:")
print(A_mult)

# Print the shape of the new one.
print("let's see if this changes its shape:")
A_mult.shape
```

```python
print("Get the first layer. In this case, the first matrix") 
print(A_mult[0,:,:])

print("Get the matrix composed from the second lines of all layers")
print(A_mult[:,1,:])

print("Get the third element of the second row from all layers.") 
print(A_mult[:,1,2])

print("Get the second element of the first row from all layers.")
print(A_mult[:,0,1])
```

```python
# Glue two copies of  "A_mult"  after each other to have six layers.
# For three dimensinal arrays, vstack concatenates along the first (0 ) axis. Equivalent to adding more matrices to our 3 dimensional array
# Meaning 
#it will change the shape from (3,4 ,4) to (6,4,4). 
A_mult_mult = np.vstack((A_mult,A_mult))
print(A_mult_mult)
print(np.shape(A_mult_mult))
```

```python
# What happens using "hstack"?
#For three dimensional arrays, hstack will glue things along the second (1) axis. That is, it will affect the rows of each matrix.
#Thus the shape will go from (3,4,4) to (3, 8, 4 )
A_mult_h = np.hstack((A_mult,A_mult))
print(A_mult_h)

# Horizontal: number of rows will be changed
print("Lets check its shape:")
print("Its shape is:")
print(A_mult_h.shape)
```

```python
# Apply linear algebraic operations with 3D arrays:
# multiply the previous A_mult with a vector.
# What should be the dimension of this?
w = np.array([0, 1, 1, 1])
A_mult@w

#The first row is the result of the Id @ v
#The second row is ther result of the matrix of ones @ v
#The third row is the result  A_2 @v
```

```python
# Reshape it: create a two-dimensional array of size (3,16)
#Condition, since we wont be removing any data, the new size must agree 
#with how many elements there are: hence 3*16 = 3 * 4 * 4.
#It reads row-wise and maintains this order.
# containing the elements of our original array A_mult
np.reshape(A_mult, (3, 16))
# we keep the number of layers, but each of them will be 
# transformed into a vector
```

```python
# Now you should completely flatten this.
# This will be later called "flatten"
A_vect = np.reshape(A_mult, (1, 48))
# OK, what is the shape of the result?
print("shape of flattening our matrix A_mult:")
print(np.shape(A_vect))
# Be careful: this has really one row, 
# but this not a vector; this is a matrix.
# So, reshape such that we really get a vector.
A_vect_vect = np.reshape(A_mult, (48,))
np.shape(A_vect_vect)
# Print both of them.
print("A_mult as a vector")
print(A_vect_vect)
print("A_mult as a matrix of size (1,48)")
print(A_vect)
```

```python
#Multiplication:

M1 = np.array([ [ [1, 1, 1],
               [2, 2, 2],
                [3, 3, 3] ], 
                  [ [10, 20, 30],
               [50, 60, 70],
                [90, 100, 110] ]])

M2 = np.ones((2,3,3))

print("let's check that the shape of M1 is (2,3,3):")
print(np.shape(M1))

print("M2, the first matrix should have a diagonal filled with 3's")
np.fill_diagonal(M2[0,:,:],([3,3,3]))
print(M2)
print("what is matmul(M1,M2)")
M1M2mult = np.matmul(M1,M2) 
print(M1M2mult)
print(f"the shape of this multiplication is {M1M2mult.shape}")

M3 = np.array ([[1, -1, 1],
                  [2, -2, 2],
                  [3, -3, 3]])
print("let's multiply M1 with the following matrix:")
print(M3)
print("result of matmul(M1,M3):")
M1M3mult = np.matmul(M1,M3) 
print(M1M3mult)
print(f"the shape of this multiplication is also {np.shape(M1M3mult)}")

#Let's check what happens with a vector:
vect = np.array([5,0,5])
M1vectmult = np.matmul(M1, vect)
print(f"the shape of the multiplication between M1 and a vector  is {M1vectmult.shape}")
print("Every matrix inside M1 collapses into a vector, making M1 a matrix as whole:")
print(M1vectmult)
```

# APPLICATION: HANDLING IMAGES

```python
# Get a nice image
image = np.array(Image.open("toy_cow.jpg"))
# Also, show this
fig = plt.figure(figsize=(15, 10))
imgplot = plt.imshow(image)
plt.title("nice little cow")
plt.show()
```

```python
image
```

```python
# For us, at the "computer level" this is an array 
# of the final dimension 3.
# Check this.
print("shape of the array representing our image: for each pixel, we have 3 values describing colors: red, green, blue")
print(image.shape)

# Not very much logical: would be more handy to have data 
# of different colors in different layers. 

# First "layer": red component, 
# second layer: green component,
# third "layer": blue component 
# All of them: integers in [0,255].
# Let us get them really separately: 

red_comp = image[:,:,0]
green_comp = image[:,:,1]
blue_comp = image[:,:,2]

#We need to manipulate these values, but as they are integers in [0, 255], we need to convert them to float64.
red_comp = red_comp.astype('float64')
green_comp = green_comp.astype('float64')
blue_comp = blue_comp.astype('float64')

# be very careful; addition does not work: 
# they are represented as integers in [0, 255]

SUM = red_comp + green_comp + blue_comp

# Just for checking: get colors in the pixel 
#  (2200, 1500) 

print("colors in the pixel (2200,1500)")
print(image[2200, 1500,:])

# OK: a lot of blue and light. 
```

```python
blue_comp>100
```

```python


print("For the pixel (2200,1500) let us see the sum of its color intensities")
print(SUM[2200,1500])

print("what is the value of maximum brightness?")
print(np.max(SUM))


# OH, there are no purple cows: I like better black ones.

# Change everything to black,  specifically that which has a lot of blue components.
# Set each component to zero.

green_comp[blue_comp>100] = 0
red_comp[blue_comp>100] = 0
blue_comp[blue_comp>100] = 0

# From these, compose the black cow.

black_cow = np.zeros((4000, 3000, 3))

black_cow[:, :, 0] = red_comp
black_cow[:, :, 1] = green_comp
black_cow[:, :, 2] = blue_comp

# Did we really "black out it" in the middle?
print("checking the color intensities in the middle")
print(black_cow[2200,1500,:])

# Technincal detail for the figure.
black_cow = black_cow.astype(int)
```

```python
fig = plt.figure(figsize=(15, 10))
plt.imshow(black_cow)
plt.title('nice black cow')
plt.show()
```

```python
# This is like a devil, it would be better to 
# have a nice black-white cow.
# OK, keep white/light colors.

# Those locations should be completely white,
# which is very light, i.e. has a large sum.


black_white_cow = np.zeros((4000, 3000, 3))

green_comp[SUM>500] = 255
red_comp[SUM>500] = 255
blue_comp[SUM>500] = 255


black_white_cow[:, :, 0] = red_comp
black_white_cow[:, :, 1] = green_comp
black_white_cow[:, :, 2] = blue_comp
#print(L[2000,1500])
#blue_comp[L] = 0
#L[0,5,56]
#print(blue_comp_orig[2200, 1500])
print(black_white_cow[2200,1500,:])
black_white_cow = black_white_cow.astype(int)
```

```python
fig = plt.figure(figsize=(15, 10))
plt.imshow(black_white_cow)
plt.title('nice black-white cow')
plt.show()
```

```python
# A pharmaceutical factory has 10 agents in a country, 
# who make contracts on 15 products in the 12 month of the year
# the corresponding number of contracts is collected in the 
# following array:   
agent_data = np.random.randint(12, size=(12, 15, 10)) # numbers from 0 to 11.
# number K at position (x, y, z) means that at the x th month 
# the product y was sold by agent z exactly K times
########        MONTH ,  PRODUCT, AGENT  #######################
#              (  10        4       4    ) --->   k = 100  
# in october, agent 4 sold product 4 100 times.   
print("How much product was sold altogether during the year?")
#What every agent sold every month, for every product.
S = np.sum(agent_data)
print(S)
print("Compute how much product #7 was sold by agent #5 during the year")
print(np.sum(agent_data[:, 7, 5]))

print("Compute how much products was sold by agent #9 during the year")
print(np.sum(agent_data[:, :, 9]))

print(" Which agent could sell product #3 most sucessfully ?")
#agent_data[:,3,:] contains how much of product 3 was sold by all agents, at every month.
# the shape of this is (12,10). Months are represented by rows and each agent by a different column.
# If we take a column, add all its elements, we obtain how much of product 3 each agent sold.
# Thus to agent_data[:,3,:] we must sum over rowwise: row_1 [0]  + row_2 [0] +... 
# Then S3 is a 1d array of how much each agent sold product 3 for the whole year. 
S3 = np.sum(agent_data[:, 3, :], axis = 0)
print(np.argmax(S3))

print("How many agents could achieve more than 100 contracts in July?")
#First, we check sells during july, this is agent_data[7,:,:].
#Then, by taking each column, adding their elements we obtain a vector
#of how many sales were done for july, we choose only the agents that had more than 100 sales
S7 = np.sum(agent_data[6, :, :], axis = 0)
print("How much each agent sold during July")
print(S7)
print(f"How many agents had more than 100 sales in july: {sum(S7 > 100)}")
```
