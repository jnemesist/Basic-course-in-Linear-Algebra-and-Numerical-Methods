#  MULTIDIMENSIONAL ARRAYS

```python
#Create an array with given "layers"
#The first one should be the identity matrix,
#the second one the matrix of ones,
#the third one should be -3 times the second one 
#check its shape
#change element at (2,3) in the last layer to be 5
A1 = np.eye(4)
print("Identity matrix:")
print(A1)
A2 = np.ones((4,4))
print(" Matrix of ones:")
print(A2)
A3 = -3*A2
print(" -3 * Matrix of ones:")
print(A3)
print(" Assembling a multidimensional array containing these 3 matrices")
A_mult = np.array([A1, A2, A3])
print(A_mult)
print("Its shape is :")
print(A_mult.shape)

# Get a specific element
print("How to access the elements of this array?")
print(f'Obtaining the element (3,2) of the third matrix : {A_mult[2,1,2]} ')


#Change it to be 5
A_mult[2,1,2] = 5

# Print the new one
print("we have changed this number from -3 to 5:")
print(A_mult)

# Print the shape of the new one.
print("let's see if this changes its shape:")
A_mult.shape
```

```python
print("Get the first layer. In this case, the first matrix") 
print(A_mult[0,:,:])

print("Get the matrix composed from the second lines of all layers")
print(A_mult[:,1,:])

print("Get the third element of the second row from all layers.") 
print(A_mult[:,1,2])

print("Get the second element of the first row from all layers.")
print(A_mult[:,0,1])
```

```python
# Glue two copies of  "A_mult"  after each other to have six layers.
# For three dimensinal arrays, vstack concatenates along the first (0 ) axis. Equivalent to adding more matrices to our 3 dimensional array
# Meaning 
#it will change the shape from (3,4 ,4) to (6,4,4). 
A_mult_mult = np.vstack((A_mult,A_mult))
print(A_mult_mult)
print(np.shape(A_mult_mult))
```

```python
# What happens using "hstack"?
#For three dimensional arrays, hstack will glue things along the second (1) axis. That is, it will affect the rows of each matrix.
#Thus the shape will go from (3,4,4) to (3, 8, 4 )
A_mult_h = np.hstack((A_mult,A_mult))
print(A_mult_h)

# Horizontal: number of rows will be changed
print("Lets check its shape:")
print("Its shape is:")
print(A_mult_h.shape)
```

```python
# Apply linear algebraic operations with 3D arrays:
# multiply the previous A_mult with a vector.
# What should be the dimension of this?
w = np.array([0, 1, 1, 1])
A_mult@w

#The first row is the result of the Id @ v
#The second row is ther result of the matrix of ones @ v
#The third row is the result  A_2 @v
```

```python
# Reshape it: create a two-dimensional array of size (3,16)
#Condition, since we wont be removing any data, the new size must agree 
#with how many elements there are: hence 3*16 = 3 * 4 * 4.
#It reads row-wise and maintains this order.
# containing the elements of our original array A_mult
np.reshape(A_mult, (3, 16))
# we keep the number of layers, but each of them will be 
# transformed into a vector
```

```python
# Now you should completely flatten this.
# This will be later called "flatten"
A_vect = np.reshape(A_mult, (1, 48))
# OK, what is the shape of the result?
print("shape of flattening our matrix A_mult:")
print(np.shape(A_vect))
# Be careful: this has really one row, 
# but this not a vector; this is a matrix.
# So, reshape such that we really get a vector.
A_vect_vect = np.reshape(A_mult, (48,))
np.shape(A_vect_vect)
# Print both of them.
print("A_mult as a vector")
print(A_vect_vect)
print("A_mult as a matrix of size (1,48)")
print(A_vect)
```

```python
#Multiplication:

M1 = np.array([ [ [1, 1, 1],
               [2, 2, 2],
                [3, 3, 3] ], 
                  [ [10, 20, 30],
               [50, 60, 70],
                [90, 100, 110] ]])

M2 = np.ones((2,3,3))

print("let's check that the shape of M1 is (2,3,3):")
print(np.shape(M1))

print("M2, the first matrix should have a diagonal filled with 3's")
np.fill_diagonal(M2[0,:,:],([3,3,3]))
print(M2)
print("what is matmul(M1,M2)")
M1M2mult = np.matmul(M1,M2) 
print(M1M2mult)
print(f"the shape of this multiplication is {M1M2mult.shape}")

M3 = np.array ([[1, -1, 1],
                  [2, -2, 2],
                  [3, -3, 3]])
print("let's multiply M1 with the following matrix:")
print(M3)
print("result of matmul(M1,M3):")
M1M3mult = np.matmul(M1,M3) 
print(M1M3mult)
print(f"the shape of this multiplication is also {np.shape(M1M3mult)}")

#Let's check what happens with a vector:
vect = np.array([5,0,5])
M1vectmult = np.matmul(M1, vect)
print(f"the shape of the multiplication between M1 and a vector  is {M1vectmult.shape}")
print("Every matrix inside M1 collapses into a vector, making M1 a matrix as whole:")
print(M1vectmult)
```

```python
#Reshape! Suppose:
A1 = np.array([
    [1, 1, 1],
    [2, 2, 2]
])

A2 = np.array([
    [3, 3, 3],
    [4, 4, 4]
])

A3 = np.array([
    [5, 5, 5],
    [6, 6, 6]
])
#We stack them along the axis 0:
A_stack = np.stack((A1,A2,A3),axis=0)
print(A_stack)
print(f"Its shape is {A_stack.shape}")

#Imagine this represents an RGB picture with 3*2 pixels
#How can we reshape this?
print("Organize it by pixel:")
print(" R  G  B")
print(A_stack.reshape(6,3))

print("getting red colors only:")
print(A_stack[:,:,0])
```
