Combine the vectors u and v ∈ R10 such that you have (u1, v1, u2, v2, . . . , u10, v10). If possible,
use vectorized operations.

```python
#Let's create two integer vectors, so its easy to see any mistakes 

u = np.arange(10,20) # 10 numbers from 10 to 20

v = np.arange(1,11)

# Vector of ones, our goal is to have something like (u1,1,u2,1,...)
e = np.ones(20)
e[::2] = u # syntax start:stop:step

#We multiply the odd entries of the vector by v:

e[1::2] = e[1::2]*v

print(e)
```

Take a random vector of length 100 and compute the vector vnew, which is the difference of the
neighboring elements with vnew[j] = v[j] − v[j − 1] for j > 0 and vnew[0] = v[0].

```python
#Let us create a random vector of size 100
v = np.random.rand(100)
#We must make the first entry equal 0:
#v [ 0] = 0
print(v)
```

```python
#Let's initialize v_new using v
v_new = np.copy(v)
# we are substracting [v_1 v_2 ... v_99] - [v_0 v_1 ... v_98] = [v_1-v_0, v_2-v_1, ..., v_99-v_98]
v_new[1:] = v[1::] - v[0:99:] 
#The first entry must be zero
v_new[0] = 0 
print(v_new)
```

(D) Take a rectangular grid with $5\times 6$ grid points
and a numbering of them $x_1, x_2, \dots, x_{30}$.
Construct a matrix $A$ such that 
$A[i,j] = 
\begin{cases} 
-4 \quad \text{if}\; x_i = x_j,\\
1  \quad \text{if}\; x_i, x_j\;\text{are vertically or horizontally neighboring,}\\
0  \quad \text{otherwise.}
\end{cases}$ 
You may use the concept of Kronecker product.

```python
#Let us build first our auxiliary matrices:
n_x = 5 # number of rows in grid
n_y = 6 # number of columns in grid
N = n_x * n_y # total number of points in grid
I_x = np.eye(n_x) #Identity matrix of size n_x
I_y = np.eye(n_y) #Identity matrix of size n_y

# To build diagonal matrices, we can use the function np.diag
# inputs : np.diag (v, k), v is the vector that will be placed in the diagonal
# k is the diagonal index, k=0 main diagonal, k>0 upper diagonals, k<0 lower diagonals
B = np.diag(-4 * np.ones(n_y), k=0) + np.diag( np.ones(n_y-1), k= -1) + np.diag( np.ones(n_y-1), k=1)
print(B)


```

```python
#Next, we will place each block using the Kron product. Let's start building a matrix C which will help 
#us place the off diagonal blocks.
C = np.diag( np.ones(n_x-1), k= -1) +np.diag( np.ones(n_x-1), k=1) 
print(C)
```

```python
#Now, time to use kron product
#Let us notice that our big block matrix A contains 5 rows and 5 columns of blocks, each block is a matrix of size 6 x 6.
# np.kron(C, I_y) places blocks of indentity matrices of size n_y *n_y in the off diagonal positions.
# np.kron(I_x,B) places the matrix B in the main diagonal position. 
A = np.kron(C, I_y) + np.kron(I_x,B) 
#Let's check the shape, it should be 30 x 30
print(np.shape(A))
#Let's see its shape
print(A)
```

```python
plt.spy(A, markersize=2)
plt.title("Sparsity pattern of A")
plt.xlabel("Column index")
plt.ylabel("Row index")
plt.show()
```
